```python
# Assignment 1
# Implement a Perceptron from Scratch
# o Implement a single-layer perceptron using NumPy.
# o Train it on linearly separable data (e.g., AND/OR gate or
# synthetic dataset).
# o Visualize decision boundary.
```

```python
# Cell 1: Import libraries and create dataset

import numpy as np
import matplotlib.pyplot as plt

X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,0,0,1])
```

```python
# Cell 2: Implement and train Perceptron

w = np.zeros(2)
b = 0
lr = 0.1

for epoch in range(10):
    for i in range(4):
        output = 1 if np.dot(X[i],w) + b >= 1 else 0
        error = y[i] - output
        w += lr * error * X[i]
        b += lr * error

print("Predictions: ",[1 if np.dot(x,w) + b >= 1 else 0 for x in X])
print("Actual: ",y)
```

```python
# Cell 3: Visualize decision boundary

plt.scatter(X[:,0], X[:,1], c=y, cmap='bwr', s=100)
x = np.linspace(-0.5,1.5,100)
plt.plot(x, (1 - b - w[0]*x) / w[1])
plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.title("Perceptron Decision Boundary (AND Gate)")
plt.grid()
plt.show()
```
