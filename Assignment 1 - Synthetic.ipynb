```python
# Implement a Perceptron from Scratch
# o Implement a single-layer perceptron using NumPy.
# o Train it on linearly separable data (e.g. synthetic dataset).
# o Visualize decision boundary.
```

```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv("dataset/perceptron.csv")

X = df[['X1', 'X2']].values
y = df['Output'].values

print("Features:", X.shape)
print("Labels:", y)
```

```python
# Implement and train Perceptron from scratch
w = np.zeros(2)
b = 0
lr = 0.1

for epoch in range(20):
    for i in range(len(X)):
        pred = 1 if np.dot(X[i], w) + b >= 0.5 else 0
        error = y[i] - pred
        w += lr * error * X[i]
        b += lr * error

predictions = (np.dot(X, w) + b >= 0.5).astype(int)

print("Actual:", y)
print("Predicted:", predictions)
print("Accuracy:", np.mean(y == predictions) * 100, "%")
```

```python
# Visualize decision boundary
plt.scatter(X[:,0], X[:,1], c=y, cmap='bwr')

x = np.linspace(0, 1, 100)
plt.plot(x, (0.5 - b - w[0]*x) / w[1], 'k')

plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Perceptron Decision Boundary")
plt.show()
```
